```python
from pyspark.sql import SparkSession
import getpass 
username=getpass.getuser()
spark=SparkSession. \
    builder. \
    config('spark.ui.port','0'). \
    config("spark.sql.warehouse.dir", f"/user/{username}/warehouse"). \
    config('spark.shuffle.useOldFetchProtocol', 'true'). \
    enableHiveSupport(). \
    master('yarn'). \
    getOrCreate()
```

```python
orders_rdd = spark.sparkContext.textFile("/public/trendytech/orders/orders_1gb.csv")
```

```python
orders_rdd.take(10)
```

```python
mapped_orders = orders_rdd.map(lambda x:(x.split(",")[2], x.split(",")[3]))
```

```python
mapped_orders.take(10)
```

```python
customers_rdd = spark.sparkContext.textFile("/public/trendytech/retail_db/customers/part-00000")
```

```python
customers_rdd.take(10)
```

```python
mapped_customers = customers_rdd.map(lambda x:(x.split(",")[0],x.split(",")[8]))
```

```python
mapped_customers.take(10)
```

```python
joined_rdd = mapped_customers.join(mapped_orders)
```

```python
joined_rdd.take(10)
```
