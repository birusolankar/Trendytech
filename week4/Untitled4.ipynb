```python
#count the orders under each status
mapped_rdd = orders_rdd.map(lambda x:(x.split(",")[3],1))
```

```python
mapped_rdd.take(10)
```

```python
aggregated_rdd = mapped_rdd.reduceByKey(lambda x,y:x+y)
```

```python
aggregated_rdd.take(10)
```

```python
sorted_rdd = aggregated_rdd.sortBy(lambda x:x[1],False)
```
